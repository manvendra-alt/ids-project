```python
import sys; sys.path.append("../src")
from data_prep import prepare_data
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

d = prepare_data("../data/KDDTrain.txt", "../data/KDDTest.txt")
print(d["X_train"].shape, d["X_val"].shape, d["X_test"].shape)
```

```python
clf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
clf.fit(d["X_train"], d["y_train"])

for name in ["val", "test"]:
    preds = clf.predict(d[f"X_{name}"])
    print("=====", name)
    print(classification_report(d[f"y_{name}"], preds, target_names=["benign", "attack"], digits=3))
    print(confusion_matrix(d[f"y_{name}"], preds))
```
